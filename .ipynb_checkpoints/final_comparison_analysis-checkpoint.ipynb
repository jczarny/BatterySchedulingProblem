# Porównanie metaheurystyk

Ten notatnik zawiera tabele, wykresy oraz grafiki wykorzystywane w pracy do porównania sześciu zaimplementowanych metaheurystyk: **ACO, HPSO, TS, GA, 2PGA i BRKGA-R-LS**. Nie uruchamia on ponownie algorytmów optymalizacyjnych, a jedynie przetwarza rezultaty wygenerowane wcześniej przez skrypt `run_final_comparison.py`. 

Analiza obejmuje:

0. charakterystykę instancji testowych oraz konfiguracji skryptu analitycznego;
1. ocenę jakości rozwiązań po ustalonym pełnym limicie czasu, z zastosowaniem ARPD, mediany i kwartyli RPD, IQR, odchylenia standardowego, częstości osiągania BKS oraz progów 0,1% i 1% od BKS;
2. weryfikację dynamiki poprawy rozwiązań w czasie oraz porównanie metaheurystyk w zestawieniu z algorytmami naiwnymi; 
3. ocenę efektywności przeszukiwania przestrzeni rozwiązań w zestawieniu z wykonaną liczbą ocen funkcji celu;
4. identyfikację obszarów przewagi poszczególnych algorytmów, uzależnionych od rozmiaru instancji i klasy obciążenia mocowego;
5. porównanie skuteczności algorytmów 2PGA oraz BRKGA-R-LS z bazowym GA.

Notatnik ten domyślnie wybiera najnowsze wygenerowane rezultaty eksperymentu z katalogu `comparison_results/final_comparison`. Notebook ten przygotowuje materiały wizualne oraz liczbowe, natomiast szczegółowa analiza i wynikające z niej wnioski przedstawione są w pracy magisterskiej. 

```python
from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm, Normalize
from matplotlib.lines import Line2D
from IPython.display import display, Markdown

warnings.filterwarnings("ignore", category=RuntimeWarning)
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 180)

PROFILE_NAME = None
PROJECT_ROOT = Path.cwd()
RESULTS_ROOT = PROJECT_ROOT / "comparison_results" / "final_comparison"
PROFILES_ROOT = RESULTS_ROOT / "profiles"

ALGORITHM_ORDER = ["ACO", "HPSO", "TS", "GA", "2PGA", "BRKGA-R-LS"]
SIZE_ORDER = [20, 50, 100, 200, 500]
PRESSURE_ORDER = ["low", "medium", "high"]
PRESSURE_LABELS = {"low": "niskie", "medium": "średnie", "high": "wysokie"}
COLORS = {
    "ACO": "#00CCFF",
    "HPSO": "#A559AA",
    "TS": "#59A89C",
    "GA": "#E02B35",
    "2PGA": "#F0C571",
    "BRKGA-R-LS": "#082A54",
}

plt.rcParams.update({
    "figure.dpi": 110,
    "savefig.dpi": 300,
    "font.family": "DejaVu Serif",
    "font.size": 10,
    "axes.titlesize": 11,
    "axes.labelsize": 10,
    "legend.fontsize": 8.5,
    "axes.grid": True,
    "grid.alpha": 0.25,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

def read_json(path):
    with path.open("r", encoding="utf-8") as handle:
        return json.load(handle)

def load_csv(name):
    path = EXPORTS_DIR / name
    if not path.exists():
        raise FileNotFoundError(f"Brakuje wymaganego eksportu: {path}")
    return pd.read_csv(path)

def ordered(frame, column="algorithm"):
    result = frame.copy()
    result[column] = pd.Categorical(result[column], ALGORITHM_ORDER, ordered=True)
    return result.sort_values(column)

def save_figure(fig, stem):
    fig.savefig(FIGURES_DIR / f"{stem}.png", dpi=300, bbox_inches="tight")
    fig.savefig(FIGURES_DIR / f"{stem}.pdf", bbox_inches="tight")
    plt.show()

def save_table(frame, stem):
    frame.to_csv(TABLES_DIR / f"{stem}.csv", index=False, encoding="utf-8-sig")
    try:
        frame.to_latex(
            TABLES_DIR / f"{stem}.tex",
            index=False,
            float_format=lambda value: f"{value:.3f}",
            escape=True,
        )
    except Exception as exc:
        print(f"Nie zapisano wariantu LaTeX tabeli {stem}: {exc}")

def show_table(frame, formats=None):
    try:
        styled = frame.style
    except AttributeError:
        display(frame)
        return
    if formats:
        styled = styled.format(formats, na_rep="—")
    display(styled.hide(axis="index"))
    
def completed_profiles():
    rows = []
    if not PROFILES_ROOT.exists():
        return rows
    for folder in PROFILES_ROOT.iterdir():
        progress_path = folder / "progress.json"
        manifest_path = folder / "manifest.json"
        exports_path = folder / "exports" / "runs.csv"
        if not (progress_path.exists() and manifest_path.exists() and exports_path.exists()):
            continue
        progress = read_json(progress_path)
        manifest = read_json(manifest_path)
        settings = manifest.get("experiment_settings", {})
        profile_runs = pd.read_csv(exports_path, usecols=["instance_id"])
        rows.append({
            "profile": folder.name,
            "completion_pct": float(progress.get("completion_pct", 0.0)),
            "saved_raw_runs": int(progress.get("saved_raw_runs", 0)),
            "expected_raw_runs": int(progress.get("expected_raw_runs", 0)),
            "time_limit_s": float(settings.get("time_limit_s", np.nan)),
            "repetitions": int(settings.get("repetitions", 0)),
            "instance_count": int(profile_runs["instance_id"].nunique()),
            "updated_at": progress.get("updated_at_local", ""),
            "path": folder,
        })
    return rows

profiles = completed_profiles()
complete = [profile for profile in profiles if profile["completion_pct"] == 100.0]
if not complete:
    raise RuntimeError(f"Brak ukończonego profilu w {PROFILES_ROOT}")

candidates = [row for row in complete]
selected = max(candidates, key=lambda row: row["updated_at"])

PROFILE_DIR = selected["path"]
EXPORTS_DIR = PROFILE_DIR / "exports"
FIGURES_DIR = PROFILE_DIR / "figures_thesis"
TABLES_DIR = PROFILE_DIR / "tables_thesis"
FIGURES_DIR.mkdir(exist_ok=True)
TABLES_DIR.mkdir(exist_ok=True)

manifest = read_json(PROFILE_DIR / "manifest.json")
progress = read_json(PROFILE_DIR / "progress.json")
settings = manifest["experiment_settings"]

runs = load_csv("runs.csv")
instances = load_csv("instances.csv")
overall = load_csv("summary_final_overall.csv")
by_size = load_csv("summary_final_by_size.csv")
by_pressure = load_csv("summary_final_by_power_pressure.csv")
by_size_pressure = load_csv("summary_final_by_size_and_power_pressure.csv")
time_runs = load_csv("time_checkpoint_runs.csv")
eval_summary = load_csv("summary_evaluation_checkpoints.csv")
ga_pairs = load_csv("ga_extension_pairs.csv")
ga_summary = load_csv("summary_ga_extensions.csv")

for frame in (runs, time_runs, ga_pairs):
    for col in frame.columns:
        if col.startswith(("hit_", "within_", "reached_", "has_", "beats_", "strictly_", "is_")):
            if frame[col].dtype == object:
                frame[col] = frame[col].map({"True": True, "False": False, True: True, False: False})

profiles_table = pd.DataFrame(profiles).drop(columns="path").sort_values(
    ["completion_pct", "time_limit_s", "updated_at"], ascending=[False, False, False]
)
display(Markdown(f"**Wybrany profil:** `{selected['profile']}`"))
show_table(profiles_table, {"completion_pct": "{:.1f}", "time_limit_s": "{:.0f}"})
```

## 0. Weryfikacja kompletności danych

Przed przejściem do właściwego porównania metaheurystyk weryfikowana jest kompletność danych oraz prawidłowość jego konfiguracji.

Tabela pierwsza przedstawia konfigurację parametrów algorytmu wyspecyfikowaną przy uruchomieniu skryptu `run_final_comparison.py`, gdzie możemy ustalić:
- przyjęty limit czasu przetwarzania dla jednego uruchomienia;
- liczbę i rozmiary instancji testowych;
- liczbę powtórzeń wykonania algorytmu na jedną instancję;
- liczbę analizowanych metaheurystyk
- liczbę uruchomień metaheurystyk, liczoną wzorem `L. meteheurystyk x L. instancji x L. powtórzęń wykonania algorytmu na instancję`;
- liczbę uruchomień algorytmów naiwnych, liczoną wzorem `L. algorytmów naiwnych x L. instasncji`;
- łączną liczbę zapisanych wyników;
- liczbę uruchomień zakończonych błędem (która musi być równa 0).

Druga część zestawienia opisuje wygenerowane instancje testowe. Dla każdego rozmiaru problemu możemy zweryfikować liczbę instancji, zakres mocy terminala, podstawowe statystyki wskaźnika obciążenia mocowego oraz liczebność klas niskiego, średniego i wysokiego obciążenia.

Uzpełnieniem tabel są wykresy przedstawiające rozkład wskaźnika obciążenia mocowego w zbiorze oraz liczebność klas obciążenia na każdy z rozmiarów instancji.

```python
metaheuristic_runs = runs.loc[~runs["is_baseline"].astype(bool)].copy()
baseline_runs = runs.loc[runs["is_baseline"].astype(bool)].copy()

overview = pd.DataFrame({
    "Parametr": [
        "Profil", "Limit czasu [s]", "Liczba instancji", "Rozmiary instancji",
        "Powtórzenia na parę algorytm–instancja", "Liczba metaheurystyk",
        "Uruchomienia metaheurystyk", "Uruchomienia algorytmów naiwnych",
        "Łączna liczba zapisanych wyników", "Liczba błędów"
    ],
    "Wartość": [
        selected["profile"], f"{settings["time_limit_s"]:.2f}", instances["instance_id"].nunique(),
        ", ".join(map(str, sorted(instances["size"].unique()))), settings["repetitions"],
        metaheuristic_runs["algorithm"].nunique(), len(metaheuristic_runs), len(baseline_runs), len(runs),
        int((runs["status"] != "ok").sum()) if "status" in runs else 0,
    ],
})
show_table(overview)

pressure_counts = (
    instances.pivot_table(index="size", columns="power_pressure_class", values="instance_id", aggfunc="count", fill_value=0)
    .reindex(index=SIZE_ORDER, columns=PRESSURE_ORDER, fill_value=0)
)
instance_table = (
    instances.groupby("size", as_index=False)
    .agg(
        liczba_instancji=("instance_id", "nunique"),
        moc_terminala_min=("terminal_max_power", "min"),
        moc_terminala_max=("terminal_max_power", "max"),
        srednie_obciazenie_mocy=("mean_power_pressure_ratio", "mean"),
        min_obciazenie_mocy=("mean_power_pressure_ratio", "min"),
        max_obciazenie_mocy=("mean_power_pressure_ratio", "max"),
    )
)
for key in PRESSURE_ORDER:
    instance_table[f"klasa_{PRESSURE_LABELS[key]}"] = instance_table["size"].map(pressure_counts[key]).fillna(0).astype(int)

instance_table_display = instance_table.rename(columns={
    "size": "n",
    "liczba_instancji": "Instancje",
    "moc_terminala_min": "Min. moc terminala",
    "moc_terminala_max": "Maks. moc terminala",
    "srednie_obciazenie_mocy": "Śr. wskaźnik obciążenia",
    "min_obciazenie_mocy": "Min. wskaźnik obciążenia",
    "max_obciazenie_mocy": "Maks. wskaźnik obciążenia",
    "klasa_niskie": "Niskie",
    "klasa_średnie": "Średnie",
    "klasa_wysokie": "Wysokie",
})
show_table(instance_table_display, {
    "Min. moc terminala": "{:.1f}", "Maks. moc terminala": "{:.1f}",
    "Śr. wskaźnik obciążenia": "{:.3f}", "Min. wskaźnik obciążenia": "{:.3f}",
    "Maks. wskaźnik obciążenia": "{:.3f}",
})
save_table(instance_table_display, "00_charakterystyka_instancji")
```

```python
fig1, ax1 = plt.subplots(figsize=(6, 4.2))

ratio_data = [
    instances.loc[
        instances["size"] == size,
        "mean_power_pressure_ratio"
    ].dropna()
    for size in SIZE_ORDER
]
parts = ax1.boxplot(
    ratio_data,
    tick_labels=SIZE_ORDER,
    patch_artist=True,
    showmeans=True
)

for box in parts["boxes"]:
    box.set_facecolor("#A0CBE8")
    box.set_alpha(0.85)

ax1.set_xlabel("Liczba akumulatorów n")
ax1.set_ylabel("Średni moc początkowa akumulatora / mocy terminala")
fig1.tight_layout()
save_figure(fig1, "00_charakterystyka_instancji_boxplot")



fig2, ax2 = plt.subplots(figsize=(7, 4.2))
bottom = np.zeros(len(SIZE_ORDER))
class_colors = {
    "low": "#59A14F",
    "medium": "#F28E2B",
    "high": "#E15759"
}

for pressure in PRESSURE_ORDER:
    values = pressure_counts[pressure].to_numpy()

    ax2.bar(
        np.arange(len(SIZE_ORDER)),
        values,
        bottom=bottom,
        label=PRESSURE_LABELS[pressure],
        color=class_colors[pressure]
    )

    bottom += values

ax2.set_xticks(np.arange(len(SIZE_ORDER)), SIZE_ORDER)
ax2.set_xlabel("Liczba akumulatorów n")
ax2.set_ylabel("Liczba instancji")

ax2.legend(
    title="Obciążenie",
    loc="center left",
    bbox_to_anchor=(1.02, 0.5)
)

fig2.tight_layout()
save_figure(fig2, "00_charakterystyka_instancji_obciazenie")
```

## 1. Jakość końcowa po ustalonym limicie czasu

Z uwagi na rozmiar i złożoność problemu, niemożliwym jest odnalezienie globalnego optimum w akceptowalnym czasie. Z tego powodu punktem odniesienia dla każdej z instancji jest najlepsze znane rozwiązanie (**BKS**, ang. *Best-Known Solution*), czyli najlepszy uzyskany wynik z dowolnego z algorytmów we wszystkich jego uruchomieniach na danej instacji. 

Na podstawie BKS liczymy następnie RPD. Wartość `RPD = 0` oznacza osiągnięcie najlepszego rozwiązania ze wszystkich algorytmów, osiągnięcie BKS na instancji oraz zwycięstwo tego algorytmu. W przypadkach remisowych (gdy rozwiązanie mieści się w tolerancji `1e-9`) obydwa rozwiązania są uznane za zwycięskie.

Na podstawie wartości RPD liczone są pozostałe statystyki:
- ARPD, które są średnią ze wszystkich osiągniętych rezultatów dla danego algorytmu;
- pierwszy, drugi i trzeci kwartyl RPD;
- rozstęp międzykwartylowy (IQR) oraz odchylenie standardowe RPD;
- liczba i odsetek uruchomień które osiągnęły BKS dla każdego z algorytmów;
- liczba i odsetek wyników osiągających rezultaty maksymalnie 1% i 0.1% gorsze od znalezionego BKS.

Jako statystyka pomocnicza weryfikowano również prędkość przetwarzania, wyrażona w liczbie wykonanych ocen rozwiązań na jednostkę czasu.

Poniższe zestawienia prezentują wyniki zbiorze algorytmów, a następnie rozkłady RPD z podziałem na rozmiary instancji.

```python
final_quality = ordered(overall)[[
    "algorithm", "arpd", "median_rpd", "q1_rpd", "q3_rpd", "iqr_rpd", "std_rpd",
    "bks_hits", "bks_hit_rate_pct", "within_0_1pct_runs", "within_0_1pct_rate_pct",
    "within_1pct_runs", "within_1pct_rate_pct"
]].rename(columns={
    "algorithm": "Algorytm", "arpd": "ARPD [%]", "median_rpd": "Mediana RPD [%]",
    "q1_rpd": "Q1 RPD [%]", "q3_rpd": "Q3 RPD [%]", "iqr_rpd": "IQR RPD [p.p.]",
    "std_rpd": "Odch. std. RPD", "bks_hits": "BKS [liczba]",
    "bks_hit_rate_pct": "BKS [%]", "within_0_1pct_runs": "Próg 0,1% [liczba]",
    "within_0_1pct_rate_pct": "Próg 0,1% [%]", "within_1pct_runs": "Próg 1% [liczba]",
    "within_1pct_rate_pct": "Próg 1% [%]",
})
show_table(final_quality, {
    "ARPD [%]": "{:.3f}", "Mediana RPD [%]": "{:.3f}", "Q1 RPD [%]": "{:.3f}",
    "Q3 RPD [%]": "{:.3f}", "IQR RPD [p.p.]": "{:.3f}", "Odch. std. RPD": "{:.3f}",
    "BKS [%]": "{:.1f}", "Próg 0,1% [%]": "{:.1f}", "Próg 1% [%]": "{:.1f}",
})
save_table(final_quality, "01_jakosc_koncowa")

final_efficiency = ordered(overall)[[
    "algorithm", "within_0_1pct_rate_pct", "median_time_to_0_1pct_s_among_successes",
    "not_reached_0_1pct_runs", "median_evaluations_to_0_1pct_among_successes",
    "within_1pct_rate_pct", "median_time_to_1pct_s_among_successes",
    "not_reached_1pct_runs", "median_evaluations_to_1pct_among_successes",
    "median_solution_count", "median_solutions_per_second",
]].rename(columns={
    "algorithm": "Algorytm", "within_0_1pct_rate_pct": "Próg 0,1% [%]",
    "median_time_to_0_1pct_s_among_successes": "Mediana czasu do 0,1% [s]",
    "not_reached_0_1pct_runs": "Nie osiągnięto 0,1% [liczba]",
    "median_evaluations_to_0_1pct_among_successes": "Mediana ocen do 0,1%",
    "within_1pct_rate_pct": "Próg 1% [%]",
    "median_time_to_1pct_s_among_successes": "Mediana czasu do 1% [s]",
    "not_reached_1pct_runs": "Nie osiągnięto 1% [liczba]",
    "median_evaluations_to_1pct_among_successes": "Mediana ocen do 1%",
    "median_solution_count": "Mediana liczby ocen",
    "median_solutions_per_second": "Mediana ocen/s",
})
show_table(final_efficiency, {
    "Próg 0,1% [%]": "{:.1f}", "Mediana czasu do 0,1% [s]": "{:.4f}",
    "Mediana ocen do 0,1%": "{:,.0f}", "Próg 1% [%]": "{:.1f}",
    "Mediana czasu do 1% [s]": "{:.4f}", "Mediana ocen do 1%": "{:,.0f}",
    "Mediana liczby ocen": "{:,.0f}",
    "Mediana ocen/s": "{:,.1f}",
})
save_table(final_efficiency, "01_efektywnosc_koncowa")
```

```python
plot_data = ordered(overall).reset_index(drop=True)
x = np.arange(len(plot_data))
med = plot_data["median_rpd"].to_numpy(float)
q1 = plot_data["q1_rpd"].to_numpy(float)
q3 = plot_data["q3_rpd"].to_numpy(float)

fig, ax = plt.subplots(figsize=(9.5, 4.8))
for idx, row in plot_data.iterrows():
    algorithm = str(row["algorithm"])
    ax.errorbar(
        idx, row["median_rpd"],
        yerr=[[row["median_rpd"] - row["q1_rpd"]], [row["q3_rpd"] - row["median_rpd"]]],
        fmt="o", markersize=7, capsize=5, linewidth=2, color=COLORS[algorithm]
    )
    ax.scatter(idx, row["arpd"], marker="D", s=38, color=COLORS[algorithm], edgecolor="black", linewidth=0.5)
ax.set_xticks(x, plot_data["algorithm"].astype(str))
ax.set_ylabel("RPD [%]")
ax.legend(handles=[
    Line2D([0], [0], marker="o", color="black", linestyle="None", label="mediana; odcinek: Q1–Q3"),
    Line2D([0], [0], marker="D", color="black", linestyle="None", label="ARPD"),
], loc="upper left")
fig.tight_layout()
save_figure(fig, "01_jakosc_koncowa_rpd")
```

```python
for size in SIZE_ORDER:
    fig, ax = plt.subplots(figsize=(6, 4.8))

    group = metaheuristic_runs.loc[
        metaheuristic_runs["size"] == size
    ]

    values = [
        group.loc[
            group["algorithm"] == alg,
            "rpd"
        ].dropna().to_numpy()
        for alg in ALGORITHM_ORDER
    ]

    parts = ax.violinplot(
        values,
        positions=np.arange(len(ALGORITHM_ORDER)),
        widths=0.85,
        showmeans=False,
        showmedians=True,
        showextrema=False,
        quantiles=[[0.25, 0.75]] * len(ALGORITHM_ORDER)
    )

    for body, algorithm in zip(parts["bodies"], ALGORITHM_ORDER):
        body.set_facecolor(COLORS[algorithm])
        body.set_edgecolor("black")
        body.set_alpha(0.75)

    parts["cmedians"].set_color("black")
    parts["cquantiles"].set_color("black")
    parts["cquantiles"].set_linewidth(0.8)

    ax.set_title(f"n = {size}")
    ax.set_xticks(
        np.arange(len(ALGORITHM_ORDER)),
        ALGORITHM_ORDER,
        rotation=60,
        ha="right"
    )

    ax.set_xlabel("")
    ax.set_ylabel("RPD [%]")

    fig.tight_layout()

    save_figure(fig, f"02_violin_rpd_n_{size}")
    plt.show()
    plt.close(fig)
```

## 2. Dynamika poprawy rozwiązania w funkcji czasu i skuteczność algorytmów w zestawieniu z algorytmami naiwnymi

Pierwszy wykres ukazuje zmianę średniej wartości RPD w kolejnych punktach w czasie. Zacieniowany obszar przylegający do punktów przedstawia przedział między pierwszym a trzecim kwartylem uzyskanch wyników. 

Drugi wykres przedstawia odsetek uruchomień który w danym momencie czasu osiągnął wynik lepszy od lepszego z dwóch algorytmów naiwnych: `LPT i SIP`. 


Zachowane punkty czasu wynoszą: `1, 5, 15, 30, 60, 120 i 150` sekund. Seria wykresów niżej przedstawia zmianę RPD w czasie z podziałem na rozmiary instancji.

```python
def aggregate_time_profiles(frame):
    return (
        frame.loc[frame["has_solution"].astype(bool)]
        .groupby(["algorithm", "checkpoint_s"], as_index=False)
        .agg(
            observations=("rpd", "count"),
            arpd=("rpd", "mean"),
            median_rpd=("rpd", "median"),
            q1_rpd=("rpd", lambda s: s.quantile(0.25)),
            q3_rpd=("rpd", lambda s: s.quantile(0.75)),
            within_0_1pct_rate_pct=("within_0_1pct", lambda s: 100 * s.mean()),
            within_1pct_rate_pct=("within_1pct", lambda s: 100 * s.mean()),
            beats_or_ties_naive_rate_pct=("beats_or_ties_naive", lambda s: 100 * s.mean()),
            strictly_beats_naive_rate_pct=("strictly_beats_naive", lambda s: 100 * s.mean()),
        )
    )

time_overall = aggregate_time_profiles(time_runs)
checkpoints = sorted(time_overall["checkpoint_s"].unique())

x_positions = np.arange(len(checkpoints))
MARKERS = {
    algorithm: marker
    for algorithm, marker in zip(
        ALGORITHM_ORDER,
        ["o", "s", "^", "D", "v", "P", "X", "*", "<", ">"]
    )
}
fig1, ax1 = plt.subplots(figsize=(8, 5))

for algorithm in ALGORITHM_ORDER:
    data = (
        time_overall
        .loc[time_overall["algorithm"] == algorithm]
        .sort_values("checkpoint_s")
    )

    x = [
        checkpoints.index(value)
        for value in data["checkpoint_s"]
    ]

    ax1.plot(
        x,
        data["arpd"].to_numpy(float),
        marker=MARKERS[algorithm],
        markersize=6,
        linewidth=1.8,
        label=algorithm,
        color=COLORS[algorithm]
    )

    ax1.fill_between(
        x,
        data["q1_rpd"].to_numpy(float),
        data["q3_rpd"].to_numpy(float),
        color=COLORS[algorithm],
        alpha=0.12
    )

ax1.set_xticks(
    x_positions,
    [f"{value:g}" for value in checkpoints]
)

ax1.set_xlabel("Czas [s]")
ax1.set_ylabel("RPD [%]")
ax1.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5)
)

fig1.tight_layout()

save_figure(fig1, "03_zmiana_arpd_w_czasie")
plt.show()


fig2, ax2 = plt.subplots(figsize=(8, 5))
for algorithm in ALGORITHM_ORDER:
    data = (
        time_overall
        .loc[time_overall["algorithm"] == algorithm]
        .sort_values("checkpoint_s")
    )

    x = [
        checkpoints.index(value)
        for value in data["checkpoint_s"]
    ]

    ax2.plot(
        x,
        data["strictly_beats_naive_rate_pct"].to_numpy(float),
        marker=MARKERS[algorithm],
        markersize=6,
        linewidth=1.8,
        label=algorithm,
        color=COLORS[algorithm]
    )

ax2.set_xticks(
    x_positions,
    [f"{value:g}" for value in checkpoints]
)

ax2.set_xlabel("Czas [s]")
ax2.set_ylabel("Uruchomienia [%]")
ax2.set_ylim(-2, 102)
ax2.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5)
)

fig2.tight_layout()

save_figure(fig2, "03_zwyciestwa_metaheurystyk_nad_naiwnymi")

plt.show()
```

Mediana czasu jest obliczana na bazie udanych uruchomień, dlatego ważnym jest interpretacja obydwu wykresów łącznie. Krótki czas nie musi świadczyć o wysokiej skuteczności algorytmu, ponieważ może osiągać zadane progi niesatysfakcjonującą liczbę razy.

## 4. Obszary przewagi poszczególnych algorytmów uzależnione od rozmiaru instancji i obciążenia mocowego

Poniższe mapy cieplne ukazują wyniki algorytmów na dwóch płąszczyznach: według rozmiaru instancji oraz według klasy obciążenia mocowego stacji ładującej.

Analiza obejmuje:
- udział osiągnięć BKS,
- udział wyników nie gorszych niż 0,1% oraz 1% od BKS,
- rozstęp międzykwartylowy (IQR) RPD jako miarę stabilności.

```python
def draw_heatmap(ax, frame, column, column_order, value_column,
                 title, xlabel, fmt=".1f",
                 cmap="cividis", vmin=None, vmax=None):

    table = (frame.pivot(index="algorithm", columns=column, values=value_column)
        .reindex(index=ALGORITHM_ORDER, columns=column_order)
    )

    values = table.to_numpy(float)

    cmap_obj = plt.get_cmap(cmap).copy()
    cmap_obj.set_bad("#dddddd")

    norm = Normalize(vmin=vmin, vmax=vmax)

    image = ax.imshow(
        values,
        aspect="auto",
        cmap=cmap_obj,
        norm=norm
    )

    labels = [PRESSURE_LABELS.get(x, x) for x in column_order]

    ax.set_xticks(
        np.arange(len(column_order)),
        labels
    )
    ax.set_yticks(
        np.arange(len(ALGORITHM_ORDER)),
        ALGORITHM_ORDER
    )
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Algorytm")
    ax.set_xticks(
        np.arange(-0.5, len(column_order), 1),
        minor=True
    )
    ax.set_yticks(
        np.arange(-0.5, len(ALGORITHM_ORDER), 1),
        minor=True
    )
    ax.grid(
        which="minor",
        color="white",
        linewidth=1
    )
    ax.tick_params(
        which="minor",
        bottom=False,
        left=False
    )

    for i in range(values.shape[0]):
        for j in range(values.shape[1]):

            value = values[i, j]

            if np.isfinite(value):
                label = format(value, fmt)
                rgba = cmap_obj(norm(value))
                
                brightness = (
                    0.2126 * rgba[0]
                    + 0.7152 * rgba[1]
                    + 0.0722 * rgba[2]
                )
                text_color = "white" if brightness < 0.5 else "black"
            else:
                label = "—"
                text_color = "black"

            ax.text(
                j,
                i,
                label,
                ha="center",
                va="center",
                color=text_color,
                fontsize=10
            )

    return image



fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6),
    constrained_layout=True
)
im = draw_heatmap(
    axes[0],
    by_size,
    "size",
    SIZE_ORDER,
    "bks_hit_rate_pct",
    "Według liczby akumulatorów",
    "Liczba akumulatorów n",
    fmt=".1f",
    cmap="cividis",
    vmin=0,
    vmax=100
)
draw_heatmap(
    axes[1],
    by_pressure,
    "power_pressure_class",
    PRESSURE_ORDER,
    "bks_hit_rate_pct",
    "Według klasy obciążenia mocowego",
    "Klasa obciążenia mocowego",
    fmt=".1f",
    cmap="cividis",
    vmin=0,
    vmax=100
)
fig.colorbar(
    im,
    ax=axes,
    shrink=0.85,
    label="Udział uruchomień [%]"
)
fig.suptitle(
    "Osiągnięcia najlepszego znanego rozwiązania (BKS)",
    fontsize=14
)

save_figure(fig, "08a_osiagniecia_bks")
plt.show()



fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6),
    constrained_layout=True
)

im = draw_heatmap(
    axes[0],
    by_size,
    "size",
    SIZE_ORDER,
    "within_0_1pct_rate_pct",
    "Według liczby akumulatorów",
    "Liczba akumulatorów n",
    fmt=".1f",
    cmap="cividis",
    vmin=0,
    vmax=100
)
draw_heatmap(
    axes[1],
    by_pressure,
    "power_pressure_class",
    PRESSURE_ORDER,
    "within_0_1pct_rate_pct",
    "Według klasy obciążenia mocowego",
    "Klasa obciążenia mocowego",
    fmt=".1f",
    cmap="cividis",
    vmin=0,
    vmax=100
)
fig.colorbar(
    im,
    ax=axes,
    shrink=0.85,
    label="Udział uruchomień [%]"
)
fig.suptitle(
    "Wyniki w odległości do 0,1% od BKS",
    fontsize=14
)
save_figure(fig, "08b_wyniki_prog_0_1pct")
plt.show()



fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6),
    constrained_layout=True
)
im = draw_heatmap(
    axes[0],
    by_size,
    "size",
    SIZE_ORDER,
    "within_1pct_rate_pct",
    "Według liczby akumulatorów",
    "Liczba akumulatorów n",
    fmt=".1f",
    cmap="cividis",
    vmin=0,
    vmax=100
)
draw_heatmap(
    axes[1],
    by_pressure,
    "power_pressure_class",
    PRESSURE_ORDER,
    "within_1pct_rate_pct",
    "Według klasy obciążenia mocowego",
    "Klasa obciążenia mocowego",
    fmt=".1f",
    cmap="cividis",
    vmin=0,
    vmax=100
)
fig.colorbar(
    im,
    ax=axes,
    shrink=0.85,
    label="Udział uruchomień [%]"
)
fig.suptitle(
    "Wyniki w odległości do 1% od BKS",
    fontsize=14
)

save_figure(fig, "08c_wyniki_prog_1pct")
plt.show()



all_iqr = np.concatenate([
    by_size["iqr_rpd"].dropna().to_numpy(float),
    by_pressure["iqr_rpd"].dropna().to_numpy(float)
])
iqr_min = all_iqr.min()
iqr_max = all_iqr.max()

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6),
    constrained_layout=True
)

im = draw_heatmap(
    axes[0],
    by_size,
    "size",
    SIZE_ORDER,
    "iqr_rpd",
    "Według liczby akumulatorów",
    "Liczba akumulatorów n",
    fmt=".3f",
    cmap="magma_r",
    vmin=iqr_min,
    vmax=iqr_max
)

draw_heatmap(
    axes[1],
    by_pressure,
    "power_pressure_class",
    PRESSURE_ORDER,
    "iqr_rpd",
    "Według klasy obciążenia mocowego",
    "Klasa obciążenia mocowego",
    fmt=".3f",
    cmap="magma_r",
    vmin=iqr_min,
    vmax=iqr_max
)
fig.colorbar(
    im,
    ax=axes,
    shrink=0.85,
    label="IQR RPD [p.p.]"
)
fig.suptitle(
    "Stabilność wyników — IQR RPD",
    fontsize=14
)

save_figure(fig, "08d_stabilnosc_iqr_rpd")
plt.show()
```

Poniższa grafika wskazuje najefektywniejszy z algorytmów dla problemu o zadanej kombinacji rozmiaru oraz klasy obciążenia mocowego. Jeżeli dwa algorytmy uzyskały tę samą medianę, zwycięskim algorytmem jest ten z niższą wartością ARPD.

Każda komórka zawiera jednocześnie liczbę instancji na zadaną kombinację (`N`) oraz różnicę między medianą RPD zwycięskiego algorytmu do algorytmu zajmującego drugie miejsce (`Δ`).

```python
cell_ranking = (
    by_size_pressure
    .sort_values([
        "size",
        "power_pressure_class",
        "median_rpd",
        "arpd"
    ])
    .copy()
)

cell_ranking["rank"] = (
    cell_ranking
    .groupby(
        ["size", "power_pressure_class"],
        observed=True
    )
    .cumcount()
)

cell_winners = (
    cell_ranking
    .loc[cell_ranking["rank"] == 0]
    .copy()
)

runners_up = (
    cell_ranking
    .loc[
        cell_ranking["rank"] == 1,
        [
            "size",
            "power_pressure_class",
            "median_rpd"
        ]
    ]
    .rename(
        columns={
            "median_rpd": "runner_up_median_rpd"
        }
    )
)


cell_winners = cell_winners.merge(
    runners_up,
    on=["size", "power_pressure_class"],
    how="left"
)

cell_winners["margin_to_runner_up"] = (
    cell_winners["runner_up_median_rpd"]
    - cell_winners["median_rpd"]
)

winner_names = (
    cell_winners
    .pivot(
        index="size",
        columns="power_pressure_class",
        values="algorithm"
    )
    .reindex(
        index=SIZE_ORDER,
        columns=PRESSURE_ORDER
    )
)

winner_counts = (
    cell_winners
    .pivot(
        index="size",
        columns="power_pressure_class",
        values="instances"
    )
    .reindex(
        index=SIZE_ORDER,
        columns=PRESSURE_ORDER
    )
)

winner_margins = (
    cell_winners
    .pivot(
        index="size",
        columns="power_pressure_class",
        values="margin_to_runner_up"
    )
    .reindex(
        index=SIZE_ORDER,
        columns=PRESSURE_ORDER
    )
)

algorithm_codes = {
    algorithm: i
    for i, algorithm in enumerate(ALGORITHM_ORDER)
}


numeric = (
    winner_names
    .map(lambda x: algorithm_codes.get(x, np.nan))
    .to_numpy(float)
)

cmap = ListedColormap([
    COLORS[algorithm]
    for algorithm in ALGORITHM_ORDER
])

norm = BoundaryNorm(
    np.arange(-0.5, len(ALGORITHM_ORDER) + 0.5, 1),
    cmap.N
)

fig, ax = plt.subplots(
    figsize=(10, 7)
)
image = ax.imshow(
    numeric,
    aspect="auto",
    cmap=cmap,
    norm=norm,
    interpolation="nearest"
)
ax.set_xticks(
    np.arange(len(PRESSURE_ORDER)),
    [PRESSURE_LABELS[p] for p in PRESSURE_ORDER]
)
ax.set_yticks(
    np.arange(len(SIZE_ORDER)),
    SIZE_ORDER
)
ax.set_xlabel(
    "Klasa obciążenia mocowego",
    fontsize=11
)
ax.set_ylabel(
    "Liczba akumulatorów n",
    fontsize=11
)
ax.tick_params(
    axis="both",
    labelsize=10
)
ax.set_xticks(
    np.arange(-0.5, len(PRESSURE_ORDER), 1),
    minor=True
)
ax.set_yticks(
    np.arange(-0.5, len(SIZE_ORDER), 1),
    minor=True
)
ax.grid(
    which="minor",
    color="white",
    linewidth=2
)
ax.tick_params(
    which="minor",
    bottom=False,
    left=False
)

for i in range(len(SIZE_ORDER)):
    for j in range(len(PRESSURE_ORDER)):

        algorithm = winner_names.iloc[i, j]

        if pd.isna(algorithm):
            continue

        count = int(winner_counts.iloc[i, j])
        margin = winner_margins.iloc[i, j]

        color = cmap(algorithm_codes[algorithm])

        brightness = (
            0.2126 * color[0]
            + 0.7152 * color[1]
            + 0.0722 * color[2]
        )

        text_color = (
            "white"
            if brightness < 0.55
            else "black"
        )

        ax.text(
            j,
            i - 0.13,
            algorithm,
            ha="center",
            va="center",
            color=text_color,
            fontsize=10,
            fontweight="bold"
        )

        ax.text(
            j,
            i + 0.18,
            f"N={count}   Δ={margin:.3f} p.p.",
            ha="center",
            va="center",
            color=text_color,
            fontsize=8.5
        )


handles = [
    Line2D(
        [0],
        [0],
        marker="s",
        linestyle="None",
        markerfacecolor=COLORS[algorithm],
        markeredgecolor="none",
        label=algorithm,
        markersize=11
    )
    for algorithm in ALGORITHM_ORDER
]

fig.tight_layout()

save_figure(
    fig,
    "10_mapa_zwyciezcow"
)

plt.show()
```

## 5. Weryfikacja skuteczności algorytmów 2PGA oraz BRKGA-R-LS w zestawieniu z bazowym GA.

Na poniższej grafice wyniki 2PGA i BRKGA-R-LS są zestawiane z wynikami GA dla tej samej instancji i tego samego numer powtórzenia. 

Porównywana jest nie tylko średnią jakość, ale też odsetek bezpośrednich zwycięstw, zmiana częstości osiągania progów 0,1% i 1% oraz - gdy oba algorytmy osiągnęły rygorystyczny próg - oszczędność czasu i liczby ocen.

```python
ga_table = ga_summary.rename(columns={
    "variant": "Wariant", "pairs": "Pary",
    "median_rpd_improvement_over_ga": "Mediana poprawy RPD [p.p.]",
    "mean_rpd_improvement_over_ga": "Średnia poprawa RPD [p.p.]",
    "variant_win_rate_pct": "Wygrywa wariant [%]", "tie_rate_pct": "Remis [%]",
    "ga_win_rate_pct": "Wygrywa GA [%]",
    "within_0_1pct_rate_change_pp": "Zmiana progu 0,1% [p.p.]",
    "pairs_where_both_reached_0_1pct": "Oba osiągnęły 0,1% [pary]",
    "median_time_saved_to_0_1pct_s_when_both_reached": "Mediana oszczędności czasu do 0,1% [s]",
    "median_evaluations_saved_to_0_1pct_when_both_reached": "Mediana oszczędności ocen do 0,1%",
    "within_1pct_rate_change_pp": "Zmiana progu 1% [p.p.]",
    "pairs_where_both_reached_1pct": "Oba osiągnęły 1% [pary]",
    "median_time_saved_to_1pct_s_when_both_reached": "Mediana oszczędności czasu do 1% [s]",
    "median_evaluations_saved_to_1pct_when_both_reached": "Mediana oszczędności ocen do 1%",
})
show_table(ga_table, {
    "Mediana poprawy RPD [p.p.]": "{:.3f}", "Średnia poprawa RPD [p.p.]": "{:.3f}",
    "Wygrywa wariant [%]": "{:.1f}", "Remis [%]": "{:.1f}", "Wygrywa GA [%]": "{:.1f}",
    "Zmiana progu 0,1% [p.p.]": "{:+.1f}",
    "Mediana oszczędności czasu do 0,1% [s]": "{:.4f}",
    "Mediana oszczędności ocen do 0,1%": "{:,.0f}",
    "Zmiana progu 1% [p.p.]": "{:+.1f}",
    "Mediana oszczędności czasu do 1% [s]": "{:.4f}",
    "Mediana oszczędności ocen do 1%": "{:,.0f}",
})
save_table(ga_table, "05_porownanie_rozszerzen_ga")

outcomes = (
    ga_pairs.groupby(["variant", "outcome"]).size().unstack(fill_value=0)
    .reindex(index=["2PGA", "BRKGA-R-LS"], columns=["ga_win", "tie", "variant_win"], fill_value=0)
)
outcomes_pct = outcomes.div(outcomes.sum(axis=1), axis=0) * 100
outcome_labels = {"ga_win": "wygrywa GA", "tie": "remis", "variant_win": "wygrywa wariant"}
outcome_colors = {"ga_win": COLORS["GA"], "tie": "#BAB0AC", "variant_win": "#76B7B2"}

fig, ax = plt.subplots(figsize=(8, 4.5))
left = np.zeros(len(outcomes_pct))
for outcome in ["ga_win", "tie", "variant_win"]:
    values = outcomes_pct[outcome].to_numpy()
    bars = ax.barh(outcomes_pct.index, values, left=left, color=outcome_colors[outcome],
                   label=outcome_labels[outcome])
    for bar, value, start in zip(bars, values, left):
        if value >= 4:
            ax.text(start + value / 2, bar.get_y() + bar.get_height()/2, f"{value:.1f}%",
                    ha="center", va="center", fontsize=8)
    left += values
ax.set_xlim(0, 100)
ax.set_xlabel("Pary uruchomień [%]")
ax.legend(ncol=3, loc="upper center", bbox_to_anchor=(0.5, -0.15))
fig.tight_layout()
save_figure(fig, "11_rozszerzenia_ga_bilans_zwyciestw")
```

Na rysunku ostatnim zaprezentowana jest efektywność algorytmów zaawansowanych w stosunku do GA w zależności od rozmiaru instancji problemu.
